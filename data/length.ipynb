```python
import os
import librosa

# Список директорий, по которым нужно пройтись
dirs = [
    "bonus_dataset/test/wavs",
    "bonus_dataset/train/wavs",
    "bonus_dataset/val/wavs",
    "stt_dataset/test/wavs",
    "stt_dataset/train/wavs",
    "stt_dataset/val/wavs",
    "tts_dataset/test/wavs",
    "tts_dataset/train/wavs",
    "tts_dataset/val/wavs",
]

# Словарь для хранения результатов
total_durations = {}

for d in dirs:
    total_duration = 0.0
    for root, _, files in os.walk(d):
        for f in files:
            if f.lower().endswith(".wav"):
                path = os.path.join(root, f)
                try:
                    duration = librosa.get_duration(path=path)
                    total_duration += duration
                except Exception as e:
                    print(f"Ошибка при обработке {path}: {e}")
    total_durations[d] = total_duration

total = 0
# Вывод результатов
for folder, total_sec in total_durations.items():
    minutes = total_sec / 60
    hours = total_sec / 3600
    total = total + hours
    print(f"{folder}: {hours:.2f} ч")

print(f"Всего: {total:.2f} ч")
```
