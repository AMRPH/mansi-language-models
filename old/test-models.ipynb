```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

data = pd.DataFrame({'base model': ['hubert-base-ls960', 'hubert-large-ll60k', 'mms-1b', 'whisper-small', 'wav2vec2-large-xlsr-53', 'gigaam'],
                     'norm_wer': [51.88, 51.74, 45.34, 40.50, 45, 46],
                     'norm_cer': [14.54, 15.17, 13.19, 12.77, 13.2, 14.5],
                     'epochs': [56, 37, 15, 8, 20, 20],
                     'size': [94.4, 315, 963, 242, 315, 242]})
```

```python
data
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 8))

colors = {
    'hubert-base-ls960': 'blue',
    'hubert-large-ll60k': 'blue',
    'mms-1b': 'green',
    'whisper-small': 'red',
    'wav2vec2-large-xlsr-53': 'orange',
    'gigaam': 'purple'
}

fig.suptitle("Точность моделей по размеру", size=15)

for i, row in data.iterrows():
    ax.scatter(row['size'], row['norm_wer'], label=row['base model'],
               s=50,
               color=colors[row['base model']], ) 

for i, row in data.iterrows():
    ax.text(row['size']+5, row['norm_wer'], row['base model'], 
            fontsize=9, 
            color=colors[row['base model']])

base = data[data['base model'] == 'hubert-base-ls960'].iloc[0]
large = data[data['base model'] == 'hubert-large-ll60k'].iloc[0]
ax.plot([base['size'], large['size']], 
        [base['norm_wer'], large['norm_wer']], 
        color='blue', linestyle='--', alpha=0.5)

ax.set_ylabel('Norm WER, %')
ax.set_xlabel('Размер модель, M')
ax.set_xlim([0, 1000])
ax.set_ylim([0, 100])

ax.legend()

ax.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig("WER по размеру.png", dpi=300, bbox_inches='tight')
plt.show()
```
